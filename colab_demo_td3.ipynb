# Kaggle / Colab Pipeline: TD3 Only (12 Distinct Conditions)

This notebook provides **12 independent experiment cells** — exactly one cell per experimental condition.
Each cell runs its assigned condition across the **3 random seeds (0, 42, 123)** for 2,000 episodes and 600 max steps,
followed automatically by 20 deterministic evaluation episodes on the best checkpoint.

**Key features:**
- Resume-safe: Safe to re-run; skips completed runs.
- Fully isolated: Each condition can be run independently in any order.

## 1. Setup & Workspace Configuration

```python
import os
import shutil
from pathlib import Path

# Clone repository if not already in the project directory
if Path("TD3-DDPG-Car-Game").exists():
    %cd TD3-DDPG-Car-Game
elif not Path("run_experiments.py").exists():
    !git clone https://github.com/adityagangwani30/TD3-DDPG-Car-Game.git
    %cd TD3-DDPG-Car-Game

# Configure working output directories
BASE_DIR = "/kaggle/working" if Path("/kaggle").exists() else "/content"
LOGS_DIR = Path(BASE_DIR) / "logs_v2"
MODELS_DIR = Path(BASE_DIR) / "models_v2"
RESULTS_DIR = Path(BASE_DIR) / "results_v2"

for d in [LOGS_DIR, MODELS_DIR, RESULTS_DIR]:
    d.mkdir(parents=True, exist_ok=True)

# Route outputs to working directory via symlinks
repo_root = Path.cwd()
for name, target in [("logs_v2", LOGS_DIR), ("models_v2", MODELS_DIR), ("results_v2", RESULTS_DIR)]:
    link_path = repo_root / name
    if link_path.resolve() != target.resolve():
        if link_path.is_symlink() or link_path.is_file():
            link_path.unlink()
        elif link_path.is_dir():
            shutil.rmtree(link_path)
        try:
            os.symlink(target, link_path)
        except OSError:
            pass

print("Working Directory:", os.getcwd())
print("logs_v2    ->", LOGS_DIR)
print("models_v2  ->", MODELS_DIR)
print("results_v2 ->", RESULTS_DIR)
```
